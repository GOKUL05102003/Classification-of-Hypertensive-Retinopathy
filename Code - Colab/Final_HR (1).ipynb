```python
import os
import cv2
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from PIL import Image
from sklearn.metrics import confusion_matrix, roc_auc_score, f1_score, precision_score, recall_score, accuracy_score, matthews_corrcoef
import matplotlib.pyplot as plt
from tqdm import tqdm

class ImagePreprocessor:
    def __init__(self):
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.RandomHorizontalFlip(),
            transforms.RandomRotation(10),
            transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ])

    def preprocess(self, image):
        img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        img_pil = Image.fromarray(img_rgb)
        return self.transform(img_pil)

class HypertensiveRetinopathyDataset(Dataset):
    def __init__(self, root_dir, preprocessor):
        self.root_dir = root_dir
        self.preprocessor = preprocessor
        self.image_paths = []
        self.labels = []

        # Define the class order explicitly
        self.class_names = ["Normal", "Mild", "Moderate", "Severe", "Malignant"]

        for class_idx, class_name in enumerate(self.class_names):
            class_dir = os.path.join(root_dir, class_name)
            if os.path.isdir(class_dir):
                for img_name in os.listdir(class_dir):
                    if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                        self.image_paths.append(os.path.join(class_dir, img_name))
                        self.labels.append(class_idx)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        label = self.labels[idx]
        image = cv2.imread(img_path)
        image_tensor = self.preprocessor.preprocess(image)
        return image_tensor, label

class HRClassifier(nn.Module):
    def __init__(self, num_classes=5):
        super(HRClassifier, self).__init__()
        self.model = models.efficientnet_b7(weights=models.EfficientNet_B7_Weights.IMAGENET1K_V1)
        in_features = self.model.classifier[1].in_features
        self.model.classifier = nn.Sequential(
            nn.Dropout(p=0.35),
            nn.Linear(in_features, 512),
            nn.ReLU(),
            nn.Dropout(p=0.25),
            nn.Linear(512, num_classes)
        )
        for i, param in enumerate(self.model.features.parameters()):
            if i < 140:
                param.requires_grad = False

    def forward(self, x):
        return self.model(x)

class FocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=2.0, reduction="mean"):
        super(FocalLoss, self).__init__()
        self.weight = weight
        self.gamma = gamma
        self.reduction = reduction
        self.ce_loss = nn.CrossEntropyLoss(weight=weight, reduction="none")

    def forward(self, inputs, targets):
        ce_loss = self.ce_loss(inputs, targets)
        pt = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean() if self.reduction == "mean" else focal_loss.sum()

def plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies):
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Val Loss')
    plt.title('Loss over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(train_accuracies, label='Train Acc')
    plt.plot(val_accuracies, label='Val Acc')
    plt.title('Accuracy over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.tight_layout()
    plt.savefig("/content/drive/MyDrive/1/hr_training_metrics.png")
    plt.show()

def train_model(model, train_loader, val_loader, optimizer, criterion, device, num_epochs=150):
    model.to(device)
    best_acc = 0.0
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.7, patience=5, verbose=True)
    train_losses, val_losses, train_accuracies, val_accuracies = [], [], [], []

    class_names = train_loader.dataset.class_names

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        for inputs, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}"):
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

        train_loss = running_loss / len(train_loader.dataset)
        train_acc = 100.0 * correct / total
        val_loss, val_acc, specificities = calculate_class_metrics(model, val_loader, criterion, device)
        scheduler.step(val_acc)

        train_losses.append(train_loss)
        val_losses.append(val_loss)
        train_accuracies.append(train_acc)
        val_accuracies.append(val_acc)

        print(f"Epoch {epoch+1}/{num_epochs} | Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%, Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%")
        print("Class Specificities:")
        for name, spec in zip(class_names, specificities):
            print(f"{name}: {spec:.2f}%")

        if val_acc > best_acc:
            best_acc = val_acc
            torch.save(model.state_dict(), "/content/drive/MyDrive/1/final_hr_model.pth")
            print(f"New best model saved with validation accuracy: {best_acc:.2f}%")

    plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies)
    print("Training complete!")
    return best_acc

def calculate_class_metrics(model, loader, criterion, device):
    model.eval()
    num_classes = 5
    correct = 0
    total = 0
    running_loss = 0.0
    class_correct = [0] * num_classes
    class_total = [0] * num_classes
    class_fp = [0] * num_classes
    class_tn = [0] * num_classes
    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            for i in range(num_classes):
                class_total[i] += (labels == i).sum().item()
                class_correct[i] += ((predicted == i) & (labels == i)).sum().item()
                class_fp[i] += ((predicted == i) & (labels != i)).sum().item()
                class_tn[i] += ((predicted != i) & (labels != i)).sum().item()
    specificities = [(class_tn[i] / (class_tn[i] + class_fp[i] + 1e-10)) * 100 for i in range(num_classes)]
    val_loss = running_loss / len(loader.dataset)
    val_acc = 100.0 * correct / total
    return val_loss, val_acc, specificities

def evaluate_model(model, loader, criterion, device):
    model.eval()
    all_labels = []
    all_preds = []
    all_probs = []
    running_loss = 0.0

    class_names = loader.dataset.class_names

    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
            probs = torch.softmax(outputs, dim=1)
            _, predicted = torch.max(probs, 1)
            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(predicted.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    val_loss = running_loss / len(loader.dataset)
    accuracy = accuracy_score(all_labels, all_preds) * 100
    sensitivity = recall_score(all_labels, all_preds, average='macro') * 100
    precision = precision_score(all_labels, all_preds, average='macro') * 100
    f1 = f1_score(all_labels, all_preds, average='macro') * 100
    mcc = matthews_corrcoef(all_labels, all_preds) * 100
    auc = roc_auc_score(np.eye(5)[all_labels], np.array(all_probs), multi_class='ovr') * 100

    cm = confusion_matrix(all_labels, all_preds)
    FP = cm.sum(axis=0) - np.diag(cm)
    TN = cm.sum() - (FP + np.diag(cm) + cm.sum(axis=1) - np.diag(cm))
    specificity = np.mean(TN / (TN + FP + 1e-10)) * 100

    print("\n--- Final Test Evaluation Metrics ---")
    print(f"Test Loss        : {val_loss:.4f}")
    print(f"Accuracy (ACC)   : {accuracy:.2f}%")
    print(f"Sensitivity (SE) : {sensitivity:.2f}%")
    print(f"Specificity (SP) : {specificity:.2f}%")
    print(f"Precision (PR)   : {precision:.2f}%")
    print(f"F1 Score         : {f1:.2f}%")
    print(f"Matthews Corr Coef (MCC): {mcc:.2f}%")
    print(f"AUC              : {auc:.3f}")

    plt.figure(figsize=(8, 6))
    plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    plt.title("Confusion Matrix")
    plt.colorbar()
    tick_marks = np.arange(len(class_names))
    plt.xticks(tick_marks, class_names, rotation=45)
    plt.yticks(tick_marks, class_names)
    plt.tight_layout()
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.savefig("/content/drive/MyDrive/1/hr_confusion_matrix.png")
    plt.show()

    return val_loss, accuracy

if __name__ == "__main__":
    train_dir = "/content/drive/MyDrive/Dataset/Train"
    val_dir = "/content/drive/MyDrive/Dataset/Val"
    test_dir = "/content/drive/MyDrive/Dataset/Test"

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    preprocessor = ImagePreprocessor()
    train_dataset = HypertensiveRetinopathyDataset(train_dir, preprocessor)
    val_dataset = HypertensiveRetinopathyDataset(val_dir, preprocessor)
    test_dataset = HypertensiveRetinopathyDataset(test_dir, preprocessor)

    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4, pin_memory=True)
    test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=4, pin_memory=True)

    model = HRClassifier(num_classes=5)

    class_counts = [0] * 5
    for _, label in train_dataset:
        class_counts[label] += 1
    print(f"Class distribution: {class_counts}")
    print(f"Class mapping: {dict(zip(range(len(train_dataset.class_names)), train_dataset.class_names))}")

    weights = 1.0 / torch.tensor(class_counts, dtype=torch.float)
    weights = weights / weights.sum() * len(class_counts)
    weights = weights.to(device)

    criterion = FocalLoss(weight=weights, gamma=2.0)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.0002, weight_decay=5e-5)

    print("Starting training...")
    best_val_acc = train_model(model, train_loader, val_loader, optimizer, criterion, device, num_epochs=150)

    model.load_state_dict(torch.load("/content/drive/MyDrive/1/final_hr_model.pth"))
    print("Evaluating on test set...")
    test_loss, test_acc = evaluate_model(model, test_loader, criterion, device)

    torch.save(model.state_dict(), "/content/drive/MyDrive/1/final_hr_model.pth")
    print("Model saved successfully.")
```

```python
import os
import cv2
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from PIL import Image
from sklearn.metrics import confusion_matrix, roc_auc_score, f1_score, precision_score, recall_score, accuracy_score, matthews_corrcoef
from sklearn.metrics import precision_recall_curve, roc_curve, auc
import matplotlib.pyplot as plt
import matplotlib.font_manager as font_manager  # Import font_manager properly
import seaborn as sns
from tqdm import tqdm
import pandas as pd
import json
from datetime import datetime

class ImagePreprocessor:
    def __init__(self):
        # Only use test transforms to ensure consistent results
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ])

    def preprocess(self, image):
        img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        img_pil = Image.fromarray(img_rgb)
        return self.transform(img_pil)

class HypertensiveRetinopathyDataset(Dataset):
    def __init__(self, root_dir, preprocessor):
        self.root_dir = root_dir
        self.preprocessor = preprocessor
        self.image_paths = []
        self.labels = []

        # Define the class order explicitly
        self.class_names = ["Normal", "Mild", "Moderate", "Severe", "Malignant"]

        for class_idx, class_name in enumerate(self.class_names):
            class_dir = os.path.join(root_dir, class_name)
            if os.path.isdir(class_dir):
                for img_name in os.listdir(class_dir):
                    if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                        self.image_paths.append(os.path.join(class_dir, img_name))
                        self.labels.append(class_idx)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        label = self.labels[idx]
        image = cv2.imread(img_path)
        image_tensor = self.preprocessor.preprocess(image)
        return image_tensor, label, img_path.split('/')[-1]  # Return filename

class HRClassifier(nn.Module):
    def __init__(self, num_classes=5):
        super(HRClassifier, self).__init__()
        self.model = models.efficientnet_b7(weights=models.EfficientNet_B7_Weights.IMAGENET1K_V1)
        in_features = self.model.classifier[1].in_features
        self.model.classifier = nn.Sequential(
            nn.Dropout(p=0.35),
            nn.Linear(in_features, 512),
            nn.ReLU(),
            nn.Dropout(p=0.25),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        return self.model(x)

class FocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=2.0, reduction="mean"):
        super(FocalLoss, self).__init__()
        self.weight = weight
        self.gamma = gamma
        self.reduction = reduction
        self.ce_loss = nn.CrossEntropyLoss(weight=weight, reduction="none")

    def forward(self, inputs, targets):
        ce_loss = self.ce_loss(inputs, targets)
        pt = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean() if self.reduction == "mean" else focal_loss.sum()

def ensure_dir(directory):
    """Make sure the directory exists, if not create it."""
    if not os.path.exists(directory):
        os.makedirs(directory)

def plot_confusion_matrix(cm, class_names, output_dir, title='Confusion Matrix', normalize=False):
    plt.figure(figsize=(10, 8))

    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] * 100
        fmt = '.1f'
        vmax = 100
    else:
        fmt = 'd'
        vmax = np.max(cm)

    sns.heatmap(cm, annot=True, fmt=fmt, cmap='Blues', xticklabels=class_names,
                yticklabels=class_names, vmin=0, vmax=vmax)
    plt.title(title)
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()

    # Save both regular and normalized versions
    suffix = '_normalized' if normalize else ''
    plt.savefig(os.path.join(output_dir, f"confusion_matrix{suffix}.png"), dpi=300, bbox_inches='tight')
    plt.close()

def plot_roc_curves(all_labels, all_probs, class_names, output_dir):
    # One-vs-Rest ROC curves for each class
    plt.figure(figsize=(10, 8))

    # Convert to one-hot encoding for ROC computation
    y_true = np.eye(len(class_names))[all_labels]

    # Convert list of arrays to a 2D numpy array
    all_probs_array = np.array(all_probs)

    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()

    for i, class_name in enumerate(class_names):
        fpr[i], tpr[i], _ = roc_curve(y_true[:, i], all_probs_array[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
        plt.plot(fpr[i], tpr[i], lw=2, label=f'{class_name} (AUC = {roc_auc[i]:.3f})')

    # Plot the diagonal
    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curves')
    plt.legend(loc="lower right")
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "roc_curves.png"), dpi=300, bbox_inches='tight')
    plt.close()

def plot_pr_curves(all_labels, all_probs, class_names, output_dir):
    # Precision-Recall curves for each class
    plt.figure(figsize=(10, 8))

    # Convert to one-hot encoding
    y_true = np.eye(len(class_names))[all_labels]

    # Convert list of arrays to a 2D numpy array
    all_probs_array = np.array(all_probs)

    # Compute PR curve for each class
    for i, class_name in enumerate(class_names):
        precision, recall, _ = precision_recall_curve(y_true[:, i], all_probs_array[:, i])
        pr_auc = auc(recall, precision)
        plt.plot(recall, precision, lw=2, label=f'{class_name} (AUC = {pr_auc:.3f})')

    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curves')
    plt.legend(loc="lower left")
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "pr_curves.png"), dpi=300, bbox_inches='tight')
    plt.close()

def plot_per_class_metrics_bar(per_class_metrics, class_names, output_dir):
    metrics = ['sensitivity', 'specificity', 'precision', 'f1']

    # Set up the bar chart
    x = np.arange(len(class_names))
    width = 0.2

    plt.figure(figsize=(12, 8))

    # Plot each metric as a group of bars
    for i, metric in enumerate(metrics):
        values = [per_class_metrics[j][metric] for j in range(len(class_names))]
        plt.bar(x + (i - 1.5) * width, values, width, label=metric.capitalize())

    plt.ylabel('Score (%)')
    plt.title('Performance Metrics by Class')
    plt.xticks(x, class_names)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.3, axis='y')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "class_metrics_comparison.png"), dpi=300, bbox_inches='tight')
    plt.close()

def evaluate_model(model, loader, criterion, device, output_dir):
    model.eval()
    all_labels = []
    all_preds = []
    all_probs = []
    all_filenames = []
    running_loss = 0.0

    class_names = loader.dataset.class_names

    # Collect predictions
    with torch.no_grad():
        for inputs, labels, filenames in tqdm(loader, desc="Evaluating"):
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)

            probs = torch.softmax(outputs, dim=1)
            _, predicted = torch.max(probs, 1)

            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(predicted.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
            all_filenames.extend(filenames)
        all_labels = np.array(all_labels)
        all_preds = np.array(all_preds)
        all_probs = np.array(all_probs)

    # Calculate overall metrics
    val_loss = running_loss / len(loader.dataset)
    accuracy = accuracy_score(all_labels, all_preds) * 100
    macro_sensitivity = recall_score(all_labels, all_preds, average='macro') * 100
    macro_precision = precision_score(all_labels, all_preds, average='macro') * 100
    macro_f1 = f1_score(all_labels, all_preds, average='macro') * 100
    mcc = matthews_corrcoef(all_labels, all_preds) * 100

    # Try/except for AUC calculation as it might fail for some edge cases
    try:
        macro_auc = roc_auc_score(np.eye(len(class_names))[all_labels], np.array(all_probs), multi_class='ovr') * 100
    except ValueError:
        macro_auc = float('nan')

    # Calculate per-class metrics
    cm = confusion_matrix(all_labels, all_preds)
    per_class_metrics = []

    for i in range(len(class_names)):
        # True positives, false positives, true negatives, false negatives
        tp = cm[i, i]
        fp = np.sum(cm[:, i]) - tp
        fn = np.sum(cm[i, :]) - tp
        tn = np.sum(cm) - tp - fp - fn

        sensitivity = (tp / (tp + fn + 1e-10)) * 100
        specificity = (tn / (tn + fp + 1e-10)) * 100
        precision_val = (tp / (tp + fp + 1e-10)) * 100
        f1_val = (2 * tp / (2 * tp + fp + fn + 1e-10)) * 100

        per_class_metrics.append({
            'sensitivity': sensitivity,
            'specificity': specificity,
            'precision': precision_val,
            'f1': f1_val
        })

    # Calculate macro-averaged specificity
    macro_specificity = np.mean([metrics['specificity'] for metrics in per_class_metrics])

    # Create results dictionary
    results = {
        'loss': val_loss,
        'accuracy': accuracy,
        'sensitivity': macro_sensitivity,
        'specificity': macro_specificity,
        'precision': macro_precision,
        'f1': macro_f1,
        'mcc': mcc,
        'auc': macro_auc,
        'confusion_matrix': cm.tolist(),
        'per_class': per_class_metrics
    }

    # Save evaluation metrics
    print("\n--- Evaluation Metrics ---")
    print(f"Loss          : {val_loss:.4f}")
    print(f"Accuracy (ACC): {accuracy:.2f}%")
    print(f"Sensitivity   : {macro_sensitivity:.2f}%")
    print(f"Specificity   : {macro_specificity:.2f}%")
    print(f"Precision     : {macro_precision:.2f}%")
    print(f"F1 Score      : {macro_f1:.2f}%")
    print(f"MCC           : {mcc:.2f}%")
    print(f"AUC           : {macro_auc:.2f}%")

    # Per-class metrics
    print("\n--- Per-Class Metrics ---")
    for i, class_name in enumerate(class_names):
        print(f"\n{class_name}:")
        print(f"  Sensitivity: {per_class_metrics[i]['sensitivity']:.2f}%")
        print(f"  Specificity: {per_class_metrics[i]['specificity']:.2f}%")
        print(f"  Precision  : {per_class_metrics[i]['precision']:.2f}%")
        print(f"  F1 Score   : {per_class_metrics[i]['f1']:.2f}%")

    # Save metrics to JSON
    with open(os.path.join(output_dir, 'evaluation_metrics.json'), 'w') as f:
        json.dump(results, f, indent=4)

    # Generate and save confusion matrix plot
    plot_confusion_matrix(cm, class_names, output_dir, title='Confusion Matrix')
    plot_confusion_matrix(cm, class_names, output_dir, title='Normalized Confusion Matrix (%)', normalize=True)

    # Generate ROC curves
    plot_roc_curves(all_labels, all_probs, class_names, output_dir)

    # Generate Precision-Recall curves
    plot_pr_curves(all_labels, all_probs, class_names, output_dir)

    # Generate bar chart of per-class metrics
    plot_per_class_metrics_bar(per_class_metrics, class_names, output_dir)

    # Save predictions for each sample
    predictions_df = pd.DataFrame({
        'filename': all_filenames,
        'true_label': [class_names[label] for label in all_labels],
        'predicted_label': [class_names[pred] for pred in all_preds],
        'correct': np.array(all_labels) == np.array(all_preds)
    })

    # Add probability columns for each class
    for i, class_name in enumerate(class_names):
        predictions_df[f'prob_{class_name}'] = [probs[i] for probs in all_probs]

    predictions_df.to_csv(os.path.join(output_dir, 'predictions.csv'), index=False)

    return results

def generate_publication_figures(metrics, class_names, output_dir):
    """Generate high-quality figures suitable for publication"""

    # 1. Create combined metrics figure (4-panel)
    plt.figure(figsize=(16, 14))
    plt.subplots_adjust(wspace=0.3, hspace=0.4)

    # Panel 1: Confusion Matrix
    plt.subplot(2, 2, 1)
    cm = np.array(metrics['confusion_matrix'])
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] * 100
    sns.heatmap(cm_normalized, annot=True, fmt='.1f', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('A) Normalized Confusion Matrix (%)', fontsize=12, fontweight='bold')
    plt.ylabel('True Label', fontsize=10)
    plt.xlabel('Predicted Label', fontsize=10)

    # Panel 2: Per-class performance comparison
    plt.subplot(2, 2, 2)
    metrics_data = {
        'Sensitivity': [metrics['per_class'][i]['sensitivity'] for i in range(len(class_names))],
        'Specificity': [metrics['per_class'][i]['specificity'] for i in range(len(class_names))],
        'Precision': [metrics['per_class'][i]['precision'] for i in range(len(class_names))],
        'F1 Score': [metrics['per_class'][i]['f1'] for i in range(len(class_names))],
    }

    df = pd.DataFrame(metrics_data, index=class_names)
    df.plot(kind='bar', rot=45, ax=plt.gca(), width=0.8)
    plt.title('B) Performance Metrics by Class', fontsize=12, fontweight='bold')
    plt.ylabel('Score (%)', fontsize=10)
    plt.xlabel('HR Classification', fontsize=10)
    plt.grid(True, linestyle='--', alpha=0.3, axis='y')
    plt.legend(loc='upper right', fontsize=8)

    # Panel 3: Radar Chart of Overall Metrics
    plt.subplot(2, 2, 3, polar=True)
    metrics_list = ['Accuracy', 'Sensitivity', 'Specificity', 'Precision', 'F1', 'AUC']
    values = [
        metrics['accuracy'],
        metrics['sensitivity'],
        metrics['specificity'],
        metrics['precision'],
        metrics['f1'],
        metrics['auc']
    ]

    # Close the polygon by repeating the first value
    metrics_list = np.append(metrics_list, metrics_list[0])
    values = np.append(values, values[0])

    # Convert to radians and plot
    angles = np.linspace(0, 2*np.pi, len(metrics_list), endpoint=True)

    plt.polar(angles, values, 'o-', linewidth=2)
    plt.fill(angles, values, alpha=0.25)
    plt.thetagrids(angles[:-1] * 180/np.pi, metrics_list[:-1])
    plt.ylim(0, 100)
    plt.title('C) Overall Performance Metrics', fontsize=12, fontweight='bold', y=1.15)

    # Add percentage labels at each point
    for angle, value, metric in zip(angles[:-1], values[:-1], metrics_list[:-1]):
        plt.text(angle, value + 10, f'{value:.1f}%',
                horizontalalignment='center', verticalalignment='center')

    # Panel 4: Class Distribution
    plt.subplot(2, 2, 4)
    cm_diag = np.diag(cm)
    plt.bar(class_names, cm_diag)
    plt.title('D) Class Distribution in Test Set', fontsize=12, fontweight='bold')
    plt.ylabel('Number of Samples', fontsize=10)
    plt.xlabel('HR Classification', fontsize=10)
    plt.xticks(rotation=45)
    plt.grid(True, linestyle='--', alpha=0.3, axis='y')

    # Save the combined figure
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "combined_results_figure.png"), dpi=600, bbox_inches='tight')
    plt.savefig(os.path.join(output_dir, "combined_results_figure.svg"), format='svg', bbox_inches='tight')
    plt.close()

    # Additional specialist plots

    # 1. Error analysis plot - comparing sensitivity vs specificity
    plt.figure(figsize=(10, 6))
    x = np.arange(len(class_names))
    width = 0.35

    sensitivity_values = [metrics['per_class'][i]['sensitivity'] for i in range(len(class_names))]
    specificity_values = [metrics['per_class'][i]['specificity'] for i in range(len(class_names))]

    plt.bar(x - width/2, sensitivity_values, width, label='Sensitivity')
    plt.bar(x + width/2, specificity_values, width, label='Specificity')

    plt.xlabel('Class')
    plt.ylabel('Score (%)')
    plt.title('Sensitivity vs Specificity by Class')
    plt.xticks(x, class_names)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.3, axis='y')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "sensitivity_vs_specificity.png"), dpi=300, bbox_inches='tight')
    plt.close()

    # 2. Create a final summary table as an image
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.axis('tight')
    ax.axis('off')

    table_data = [
        ['Metric', 'Score (%)'],
        ['Accuracy', f"{metrics['accuracy']:.2f}"],
        ['Sensitivity', f"{metrics['sensitivity']:.2f}"],
        ['Specificity', f"{metrics['specificity']:.2f}"],
        ['Precision', f"{metrics['precision']:.2f}"],
        ['F1 Score', f"{metrics['f1']:.2f}"],
        ['MCC', f"{metrics['mcc']:.2f}"],
        ['AUC', f"{metrics['auc']:.2f}"]
    ]

    table = ax.table(cellText=table_data, loc='center', cellLoc='center')
    table.auto_set_font_size(False)
    table.set_fontsize(12)
    table.scale(1, 1.5)

    # Fix for the font manager error - Use correct approach to set bold text
    for (i, j), cell in table.get_celld().items():
        if i == 0:
            cell.set_text_props(weight='bold')  # Correct way to set bold text

    plt.title('Summary of Evaluation Metrics', fontsize=14, pad=20)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, "metrics_summary_table.png"), dpi=300, bbox_inches='tight')
    plt.close()

if __name__ == "__main__":
    # Paths
    base_dir = "/content/drive/MyDrive"
    test_dir = f"{base_dir}/Dataset/Test"
    model_path = f"{base_dir}/1/final_hr_model.pth"  # Path to your saved model

    # Create results directory with timestamp
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    results_dir = f"{base_dir}/HR_Evaluation_Results_{timestamp}"
    ensure_dir(results_dir)

    # Set device
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    # Initialize dataset and dataloader
    preprocessor = ImagePreprocessor()
    test_dataset = HypertensiveRetinopathyDataset(test_dir, preprocessor)
    test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=4, pin_memory=True)

    # Initialize model and load saved weights
    model = HRClassifier(num_classes=5)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.to(device)
    print(f"Model loaded from {model_path}")

    # Set up loss function (for loss calculation only)
    criterion = nn.CrossEntropyLoss()  # Simple loss for evaluation

    # Create evaluation directory
    eval_dir = results_dir
    ensure_dir(eval_dir)

    # Evaluate the model
    print("\nEvaluating on test set...")
    test_metrics = evaluate_model(model, test_loader, criterion, device, eval_dir)

    # Create publication-ready figures
    print("\nGenerating publication-quality figures...")
    publication_dir = os.path.join(results_dir, "publication_figures")
    ensure_dir(publication_dir)
    generate_publication_figures(test_metrics, test_dataset.class_names, publication_dir)

    # Create summary file with all key results
    with open(os.path.join(results_dir, "summary.txt"), 'w') as f:
        f.write("Hypertensive Retinopathy Classification Results Summary\n")
        f.write("="*60 + "\n\n")
        f.write(f"Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        f.write(f"Model: EfficientNet-B7\n")
        f.write(f"Classes: {', '.join(test_dataset.class_names)}\n")
        f.write(f"Test samples: {len(test_dataset)}\n\n")

        f.write("Test Set Performance:\n")
        f.write(f"Accuracy: {test_metrics['accuracy']:.2f}%\n")
        f.write(f"Sensitivity: {test_metrics['sensitivity']:.2f}%\n")
        f.write(f"Specificity: {test_metrics['specificity']:.2f}%\n")
        f.write(f"Precision: {test_metrics['precision']:.2f}%\n")
        f.write(f"F1 Score: {test_metrics['f1']:.2f}%\n")
        f.write(f"MCC: {test_metrics['mcc']:.2f}%\n")
        f.write(f"AUC: {test_metrics['auc']:.2f}%\n\n")

        f.write("Per-Class Performance:\n")
        for i, class_name in enumerate(test_dataset.class_names):
            f.write(f"\n{class_name}:\n")
            f.write(f"  Sensitivity: {test_metrics['per_class'][i]['sensitivity']:.2f}%\n")
            f.write(f"  Specificity: {test_metrics['per_class'][i]['specificity']:.2f}%\n")
            f.write(f"  Precision: {test_metrics['per_class'][i]['precision']:.2f}%\n")
            f.write(f"  F1 Score: {test_metrics['per_class'][i]['f1']:.2f}%\n")

    print("\nAll results saved to: " + results_dir)
    print("The following visualizations were generated:")
    print("1. Confusion matrices (raw counts and normalized)")
    print("2. ROC curves for each class")
    print("3. Precision-Recall curves for each class")
    print("4. Bar chart comparing metrics across classes")
    print("5. Combined 4-panel figure for publication")
    print("6. Sensitivity vs Specificity comparison")
    print("7. Summary metrics table as image")

    print("\nEvaluation complete!")
```
